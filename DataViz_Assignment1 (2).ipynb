### Exercise B2 (2 points)
Using `sns.barplot`, plot the **average tip amount** for each `day` of the week. Which day has the highest average tip?


````python
# Your Code
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
tips = sns.load_dataset('tips')

# Bar plot of average tip by day
sns.barplot(x='day', y='tip', data=tips)

plt.title('Average Tip by Day')
plt.ylabel('Average Tip ($)')
plt.xlabel('Day of the Week')
plt.show()

# To determine which day has the highest average tip:
print(tips.groupby('day')['tip'].mean().sort_values(ascending=False))

````

### Exercise C3 — Pie Chart (2 points)
Create a pie chart showing the proportion of passengers who were `alone` vs. not alone (`alone` column is boolean). Show percentages on the chart.


````python
# Your Code
import seaborn as sns
import matplotlib.pyplot as plt

# Load titanic dataset
titanic = sns.load_dataset('titanic')

# Count passengers who were alone vs. not alone
alone_counts = titanic['alone'].value_counts()

# Create pie chart with percentages
plt.figure(figsize=(6, 6))
plt.pie(alone_counts, 
        labels=['Alone', 'Not Alone'] if True in alone_counts.index else ['Not Alone', 'Alone'],
        autopct='%1.1f%%', 
        startangle=90,
        colors=['#66b3ff', '#ffcc99'])

plt.title('Proportion of Passengers Alone vs. Not Alone')
plt.show()
````

### Exercise E1 (1 point)
Take any one plot from this notebook and save it as:
1. A PNG file at 300 dpi (`my_plot.png`)
2. An SVG vector file (`my_plot.svg`)

Briefly state (in a markdown cell) which format you'd choose for a slide presentation vs. for a scientific journal figure, and why.

#### Notes
For a **slide presentation**, a PNG at a reasonable dpi (150-300) is usually sufficient since slides are viewed at a fixed size on screens/projectors, and PNG is universally supported and lightweight. For a **scientific journal figure**, an SVG (or other vector format like PDF/EPS) is preferable because it can be resized/rescaled without any loss of quality — important since journals often resize figures to fit columns and print at high resolution.
